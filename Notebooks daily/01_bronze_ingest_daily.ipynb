```python
spark.sql("CREATE CATALOG IF NOT EXISTS oil_stock")
spark.sql("USE CATALOG oil_stock")
```

```python
import yfinance as yf
import pandas as pd

tickers = ["EQNR.OL", "BZ=F"]

frames = []

for ticker in tickers:

    df = yf.download(
        ticker,
        period="5d",
        interval="1d",
        auto_adjust=False
    )

    df = df.reset_index()

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df["symbol"] = ticker

    frames.append(df)

daily = pd.concat(frames, ignore_index=True)

display(daily)
```

```python
from pyspark.sql import functions as F
from delta.tables import DeltaTable

today_oslo = F.to_date(
    F.from_utc_timestamp(
        F.current_timestamp(),
        "Europe/Oslo"
    )
)

bronze_daily_df = (
    spark.createDataFrame(daily)
    .select(
        F.col('Date').alias('trading_date'),
        F.col("symbol"),
        F.col("Open").alias("open"),
        F.col("High").alias("high"),
        F.col("Low").alias("low"),
        F.col("Close").alias("close"),
        F.col("Volume").alias("volume")
    )
    .withColumn(
        'currency',
        F.when(F.col('symbol') == 'EQNR.OL', 'NOK')
        .when(F.col('symbol') == 'BZ=F', 'USD')
    )
    .withColumn('source', F.lit('yahoo_finance'))
    .withColumn('ingested_at', F.current_timestamp())
    .filter(
            F.col("trading_date") < today_oslo
    )
)

display(bronze_daily_df)
```

```python
target = DeltaTable.forName(
    spark,
    'bronze.market_daily'
)

(
    target.alias('t')
    .merge(
        bronze_daily_df.alias('s'),
        """
        t.symbol = s.symbol
        AND t.trading_date = s.trading_date
        """
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)
```
